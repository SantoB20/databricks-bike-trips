```python
from pyspark.sql.functions import date_format, count, sum
```

```python
spark.read.table("bikes.01_bronze.chicago_trips_raw").\
    groupBy(date_format("started_at", "yyyy-MM").alias("date")).\
    agg(count("*").alias("trips")).\
    orderBy("date").display()
```

```python
%sql
select *
from bikes.`02_silver`.stations_cleansed
where station_id in (
    select station_id
    from bikes.`02_silver`.stations_cleansed
    group by station_id
    having count(*) > 1
)
order by station_id asc;
```

```python
%sql
select station_id, count(*) from bikes.`02_silver`.stations_cleansed
group by station_id
order by count(*) desc;
```
