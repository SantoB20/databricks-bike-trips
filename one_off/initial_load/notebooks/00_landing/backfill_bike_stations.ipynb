```python
import requests
import os
import shutil
from datetime import date
```

```python
data_sources: list = [
    {
        "id": "divvy",
        "url": "https://gbfs.lyft.com/gbfs/2.3/chi/en/station_information.json"
    },
    {
        "id": "baywheels",
        "url": "https://gbfs.lyft.com/gbfs/2.3/bay/en/station_information.json"
    }
]

volume_path : str = "/Volumes/bikes/00_landing/data_sources"
now = date.today().strftime("%Y-%m")
for data_source in data_sources:
    res = requests.get(data_source["url"])
    if res.status_code != 200:
        continue
    file_name: str = f"{data_source['id']}_stations_lookup.json"
    os.makedirs(f"{volume_path}/{data_source['id']}/stations/{now}", exist_ok=True)
    local_path: str = f"{volume_path}/{data_source['id']}/stations/{now}/{file_name}"
    with open(local_path, "wb") as file:
        file.write(res.content)
```
