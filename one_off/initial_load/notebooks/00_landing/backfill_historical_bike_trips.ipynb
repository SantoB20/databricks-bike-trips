```python
import requests
import os
from io import BytesIO
from zipfile import ZipFile
import shutil
```

```python
data_sources: list = [
    {
        "id": "divvy",
        "urls": [
            "https://divvy-tripdata.s3.amazonaws.com/($date)-divvy-tripdata.zip",
            "https://divvy-tripdata.s3.amazonaws.com/($date)-divvy-tripdata.csv.zip"
        ]
    },
    {
        "id": "baywheels",
        "urls": [
            "https://s3.amazonaws.com/baywheels-data/($date)-baywheels-tripdata.zip",
            "https://s3.amazonaws.com/baywheels-data/($date)-baywheels-tripdata.csv.zip"
        ]
    }
]

dates : list = [("2026","02"),("2026","03"),("2026","04"),("2026","05"),("2026","06")]
volume_path : str = "/Volumes/bikes/00_landing/data_sources"

for year, month in dates:
    for data_source in data_sources:
        for url in data_source["urls"]:
            url = url.replace("($date)", f"{year}{month}")
            res = requests.get(url)
            if res.status_code != 200:
                continue
            file_name: str = f"{year}{month}-{data_source['id']}-tripdata.csv"
            os.makedirs(f"{volume_path}/{data_source['id']}/trips/{year}-{month}", exist_ok=True)
            local_path = f"{volume_path}/{data_source['id']}/trips/{year}-{month}/{file_name}"
            with ZipFile(BytesIO(res.content)) as zip:
                with zip.open(file_name) as csv:
                    with open(local_path, "wb") as file:
                        shutil.copyfileobj(csv, file)
            break

            
```
