```python
from pyspark.sql.functions import current_timestamp, lit
```

```python
landing_sources: list = [
    {
        "id": "divvy",
        "city_id": "chicago",
        "city": "Chicago",
        "volume_path": "/Volumes/bikes/00_landing/data_sources/divvy/trips/"
    },
    {
        "id": "baywheels",
        "city_id": "san_francisco",
        "city": "San Francisco",
        "volume_path": "/Volumes/bikes/00_landing/data_sources/baywheels/trips/"
    }
]
```

```python
for landing_source in landing_sources:
    df = spark.read.format("csv").option("header",True).load(f"{landing_source['volume_path']}/*")
    df = df.withColumn("processed_timestamp", current_timestamp()).\
            withColumn("city", lit(landing_source['city']))
    df.write.mode("overwrite").saveAsTable(f"bikes.01_bronze.{landing_source['city_id']}_trips_raw")
```
