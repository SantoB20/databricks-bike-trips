```python
from pyspark.sql.functions import lit, explode
```

```python
landing_sources: list = [
    {
        "id": "divvy",
        "city_id": "chicago",
        "city": "Chicago",
        "volume_path": "/Volumes/bikes/00_landing/data_sources/divvy/stations/2026-09"
    },
    {
        "id": "baywheels",
        "city_id": "san_francisco",
        "city": "San Francisco",
        "volume_path": "/Volumes/bikes/00_landing/data_sources/baywheels/stations/2026-09"
    }
]
```

```python
for landing_source in landing_sources:
    df = spark.read.format("json").load(f"{landing_source["volume_path"]}")
    df = df.select(explode("data.stations").alias("stations")).\
            select("stations.*").\
            withColumn("city", lit(landing_source["city"]))
    df.write.mode("overwrite").saveAsTable(f"bikes.01_bronze.{landing_source["city_id"]}_station_lookup_raw")
```
